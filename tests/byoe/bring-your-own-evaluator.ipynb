# BYOE: Bring Your Own Evaluator

```python
import webbrowser

import llm_byoe_demo_evaluator

from h2o_sonar import config as h2o_sonar_config
from h2o_sonar import interpret as evaluate
from h2o_sonar.utils import testing


h2ogpte_connection = h2o_sonar_config.ConnectionConfig(
    connection_type=h2o_sonar_config.ConnectionConfigType.H2O_GPT_E.name,
    name="H2O GPT Enterprise",
    description="H2O GPT Enterprise.",
    server_url="http://h2ogpte.h2o.ai/",
    token="YOUR_API_TOKEN_HERE",
    token_use_type=h2o_sonar_config.TokenUseType.API_KEY.name,
)
```

```python
#
# my CUSTOM evaluator
#
h2o_sonar_config.config.custom_explainers = [
    "llm_byoe_demo_evaluator::BringYourOwnEvaluatorDemo"
]

# test lab
test_lab = testing.RagTestLab.load_from_json(
    rag_connection=h2ogpte_connection,
    file_path="data/generative/h2ogpte_benchmark_test_lab_top.json",
    docs_cache_dir="data/generative/h2ogpte-documents",
)

evaluation = evaluate.run_interpretation(
    # dataset: prompts, constraints and model keys
    dataset=test_lab.dataset,
    # models to be evaluated & compared to get leaderboard
    models=list(test_lab.rag_models.values()),
    # evaluators
    explainers=[
        llm_byoe_demo_evaluator.BringYourOwnEvaluatorDemo().explainer_id(),
    ],
    # where to save the report
    results_location="data/generative/bring-your-own-evaluator/report",
)

webbrowser.open(evaluation.result.get_html_report_location())
```
